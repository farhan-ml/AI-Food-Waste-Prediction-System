```python
# Import Label Encoder

from sklearn.preprocessing import LabelEncoder
```

```python
# Create Label Encoder

encoder = LabelEncoder()
```

```python
categorical_columns = ["Type of Food","Event Type","Storage Conditions","Purchase History","Seasonality","Preparation Method","Geographical Location",
    "Pricing"
]
```

```python
# Apply Label Encoding

for column in categorical_columns:

    df[column] = encoder.fit_transform(df[column])
```

```python
# Features

X = df.drop("Wastage Food Amount", axis=1)
```

```python
# Target

y = df["Wastage Food Amount"]
```

```python
# Features Shape

print(X.shape)

# Target Shape

print(y.shape)
```

```python
from sklearn.model_selection import train_test_split

# Split Dataset

X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.20, random_state=42)
```

```python
# Training Shape

print(X_train.shape)

print(y_train.shape)
```

```python
# Testing Shape

print(X_test.shape)

print(y_test.shape)
```

```python
# Import Random Forest

from sklearn.ensemble import RandomForestRegressor
# Create Model

model = RandomForestRegressor()
model.fit(X_train, y_train)
```

```python
# Prediction

y_predict = model.predict(X_test)
```

```python
df_compare = pd.DataFrame({'Actual': y_test, 'Predicted': y_predict}).reset_index(drop = True)
df_compare.head(10)
```

```python
# Import Evaluation Metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mae = mean_absolute_error(y_test, y_predict)
print('mean_absolute_erro:', mae)
mse = mean_squared_error(y_test, y_predict)
print('mean_squared_error:', mse)
r2 = r2_score(y_test, y_predict)
print('r2_score:', r2)
```

```python
# Import Gradient Boosting

from sklearn.ensemble import GradientBoostingRegressor
```

```python
# Create Model

gb_model = GradientBoostingRegressor()
gb_model.fit(X_train, y_train)
```

```python
# Prediction

gb_prediction = gb_model.predict(X_test)
```

```python
# Model Evaluation

gb_mae = mean_absolute_error(y_test, gb_prediction)

gb_rmse = np.sqrt(mean_squared_error(y_test, gb_prediction))

gb_r2 = r2_score(y_test, gb_prediction)

print("Gradient Boosting Performance")

print("-"*35)

print("MAE :", round(gb_mae,2))

print("RMSE :", round(gb_rmse,2))

print("R2 Score :", round(gb_r2,4))
```

```python
# Import XGBoost

from xgboost import XGBRegressor
```

```python
# Create Model

xgb_model = XGBRegressor()
xgb_model.fit(X_train, y_train)
```

```python
# Prediction

xgb_prediction = xgb_model.predict(X_test)
```

```python
# Model Evaluation

xgb_mae = mean_absolute_error(y_test, xgb_prediction)

xgb_rmse = np.sqrt(mean_squared_error(y_test, xgb_prediction))

xgb_r2 = r2_score(y_test, xgb_prediction)

print("XGBoost Performance")

print("-"*30)

print("MAE :", round(xgb_mae,2))

print("RMSE :", round(xgb_rmse,2))

print("R2 Score :", round(xgb_r2,4))
```

```python
# Model Comparison

comparison = pd.DataFrame({

    "Model":["Random Forest", "Gradient Boosting","XGBoost"],

    "MAE":[mae, gb_mae, xgb_mae],

    "RMSE":[mse, gb_rmse, xgb_rmse],

    "R2 Score":[r2, gb_r2, xgb_r2]

})

comparison
```
